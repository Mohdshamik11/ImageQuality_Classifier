# Phase 2b restoration training — Kaggle runner

This notebook is a thin wrapper. The real training code lives in the GitHub repo

## One-time setup (do this before the first run)

1. **Upload the clean images as a Kaggle Dataset.**
   - On your PC, zip the *contents* of `data/clean_pool_small/` so the zip root holds
     `coco/` and `div2k/` (not a `clean_pool_small/` folder).
   - Kaggle → *Datasets* → *New Dataset* → upload that zip.
   - Title it exactly **`imagequality clean pool`** → slug becomes `imagequality-clean-pool`.
2. **In this notebook** (right-hand panel):
   - *Add Input* → your `imagequality-clean-pool` dataset.
   - *Accelerator* → **GPU T4 x2** (the script uses one).
   - *Internet* → **On** (needed for `git clone` + `pip install`; may need phone verification).
3. Run all cells. A 45-epoch run is roughly **1.5–2 h**.
